1.
Вивести кількість фільмів в кожній категорії.
Результат відсортувати за спаданням.

```python
film_category_df.show()
```

```python
category_df.show()
```

```python
film_category_joined_df = film_category_df.join(
    category_df, 
    on="category_id", 
    how="inner"
)
```

```python
film_category_joined_df.show()
```

```python
films_per_category_df = film_category_joined_df.groupBy("name").agg(F.count("film_id").alias("film_count")).orderBy(F.desc("film_count"))
```

```python
films_per_category_df.show()
```

2.
Вивести 10 акторів, чиї фільми брали на прокат найбільше.
Результат відсортувати за спаданням.

```python
actor_df.show()
```

```python
film_actor_df.show()
```

```python
inventory_df.show()
```

```python
rental_df.show()
```

```python
joined_df = actor_df.join(
        film_actor_df, 
        on="actor_id", 
        how="inner"
    ).join(
        inventory_df, 
        on="film_id", 
        how="inner"
    ).join(
        rental_df, 
        on="inventory_id", 
        how="inner"
    )
```

```python
joined_df
```

```python
result = joined_df.groupBy("actor_id", "first_name", "last_name").agg(
    F.count("rental_id").alias("Rented_times")
).orderBy(F.desc("Rented_times")).limit(10)
```

```python
result.show()
```

3.
Вивести категорію фільмів, на яку було витрачено найбільше грошей
в прокаті

```python
joined_df = film_df.join(
        film_category_df, 
        on="film_id", 
        how="inner"
    ).join(
        category_df, 
        on="category_id", 
        how="inner"
    ).join(
        inventory_df, 
        on="film_id", 
        how="inner"
    ).join(
        rental_df, 
        on="inventory_id", 
        how="inner"
    ).join(
        payment_df, 
        on="rental_id", 
        how="inner"
    )
```

```python
joined_df
```

```python
result = joined_df.groupBy("name").agg(
    F.sum("amount").alias("Total_revenue")
).orderBy(F.desc("Total_revenue")).limit(1)
```

```python
result.show()
```

5.
Вивести топ 3 актори, які найбільше зʼявлялись в категорії фільмів “Children”

```python
category_df.show()
```

```python
joined_df = actor_df.join(
        film_actor_df, 
        on="actor_id", 
        how="inner"
    ).join(
        film_category_df, 
        on="film_id", 
        how="inner"
    )
```

```python
joined_df
```

```python
result_df = joined_df.filter(F.col("category_id") == 3).groupBy("actor_id", "first_name", "last_name").agg(
    F.count("film_id").alias("Appeared_in_films_of_Children_category")
).orderBy(F.desc("Appeared_in_films_of_Children_category")).limit(3)
```
